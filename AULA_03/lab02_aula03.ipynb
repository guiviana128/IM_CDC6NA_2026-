## ROS 2

```python
import rclpy
from rclpy.node import Node
from geometry_msgs.msg import Twist
import time
import math

class Giro(Node):
    def __init__(self):
        super().__init__("giro_aula03")
        self.publicador = self.create_publisher(Twist, "/cmd_vel", 10)

    def girar(self):
        angulo = math.pi / 2
        velocidade = 0.5
        tempo_giro = angulo / velocidade

        comando = Twist()
        comando.angular.z = velocidade

        inicio = time.time()

        while time.time() - inicio < tempo_giro:
            self.publicador.publish(comando)
            time.sleep(0.05)

        self.publicador.publish(Twist())

        print("Giro finalizado")

rclpy.init()
node = Giro()

try:
    node.girar()
finally:
    node.destroy_node()
    rclpy.shutdown()
```
